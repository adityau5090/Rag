```python
from langchain_community.document_loaders.json_loader import JSONLoader
from pathlib import Path
from pprint import pp
```

```python
file_path = Path("../knowledgeSource/rag_knowledge_200_lines.json")
file_path.exists()
```

```python
def metadata_function(record: dict, metadata: dict) -> dict:
    metadata["id"] = record['id']
    metadata["topic"] = record['topic']
    metadata["workflow_stage"] = record['workflow_stage']
    
    return metadata
```

```python
loader = JSONLoader(file_path,
                    jq_schema=".[]",
                    content_key="definition",
                    metadata_func=metadata_function)
```

```python
documents = loader.load()
documents
```
