# Document based splitting 

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter, Language
```

## Splitting of Python Code

```python
python_Code = """ 
class Document:

    def __init__(self, title, author, content):
        self.title = title
        self.author = author
        self.content = content

    def display_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")

    def get_content(self):
        return self.content

    def count_words(self):
        return len(self.content.split())

    def display_content(self):
        print("\nDocument Content:")
        print(self.content)

    def update_content(self, new_content):
        self.content = new_content
        print("\nDocument updated successfully!")


def main():

    # Creating document objects
    doc1 = Document(
        "Introduction to Artificial Intelligence",
        "John",
        '''
        Artificial Intelligence is a branch of computer science
        that focuses on creating intelligent machines.

        Machine Learning is a subset of Artificial Intelligence
        that allows systems to learn from data.

        Deep Learning uses neural networks to solve complex
        problems such as image recognition and natural language
        processing.
        '''
    )

    # Display document information
    doc1.display_info()

    # Display document content
    doc1.display_content()

    # Count total words
    print("\nTotal Words:", doc1.count_words())

    # Retrieve document content
    content = doc1.get_content()

    print("\nRetrieved Content:")
    print(content)

    # Update document
    doc1.update_content(
        "Generative AI can generate text, images, audio and code."
    )

    # Display updated content
    doc1.display_content()


if __name__ == "__main__":
    main()

"""
```

```python
splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.PYTHON,
    chunk_size=500,
    chunk_overlap=50
)
```

```python
chunks = splitter.split_text(python_Code)
```

```python
from termcolor import COLORS, colored
from random import choice
```

```python
def display_chunks(chunks):
    colors_list = list(COLORS.keys())[2:8]

    print(f"Total Number of Chunks: {len(chunks)}")

    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        print(colored(text=chunk, color=choice(colors_list)), end="\n\n")
```

```python
display_chunks(chunks)
```
