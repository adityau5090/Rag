## Character Based Splitting

```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
text="""Artificial Intelligence (AI) is a branch of computer science that focuses on creating machines and systems capable of performing tasks that typically require human intelligence. These tasks include learning, problem-solving, decision-making, and understanding language.

AI is becoming an important part of our daily lives. We use it in virtual assistants, recommendation systems, self-driving cars, healthcare, education, and many other fields. Technologies like ChatGPT, Google Gemini, and voice recognition systems are examples of AI applications.

Artificial Intelligence has the potential to make our lives easier, improve productivity, and solve complex problems. However, it also brings challenges such as data privacy, job displacement, and ethical concerns. Therefore, AI should be developed and used responsibly to benefit society."""
```

```python
splitter = CharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap=10,
    length_function=len,
    separator=" "
)
```

```python
chunks = splitter.split_text(text)
chunks
```

```python
from termcolor import colored, COLORS
from random import choice
```

```python
print(COLORS)
```

```python
colors_list = list(COLORS.keys())[2:8]
colors_list
```

```python
def create_chunks(text: str, chunk_size: int, seperator: str, chunk_overlap: int = 0) -> list[str]:
    splitter = CharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        separator=seperator
    )
    
    chunks = splitter.split_text(text)
    
    return chunks
```

```python
def dislpay_chunks_with_overlaps(chunks, chunk_overlap):
    if not chunk_overlap:
        raise(ValueError("Chunk overlap cannot be 0"))
    
    overlap_text = ""
    overlap_color = "white"
    colors_list = list(COLORS.keys())[2:8]
    
    print(f"Total no. of chunks : {len(chunks)}")
    for num, chunk in enumerate(chunks, 1):
        print(f"Chunk {num}: Length {len(chunk)} chars")
        if num == 1:
            non_overlap_text = chunk[0:-chunk_overlap]
            overlap_text = chunk[-chunk_overlap:]
            print(colored(non_overlap_text, choice(colors_list)),
                colored(overlap_text, overlap_color),
                sep="", end="\n\n"
            )
            
        else:
            print(colored(overlap_text,overlap_color), end="")
            non_overlap_text = chunk[chunk_overlap:-chunk_overlap]
            print(colored(non_overlap_text, choice(colors_list)), end="")
            overlap_text = chunk[-chunk_overlap:]
            print(colored(overlap_text, overlap_color), end="\n\n")
```

```python
dislpay_chunks_with_overlaps(chunks=create_chunks(text, chunk_size=100, seperator=" ", chunk_overlap=10), chunk_overlap=10)
```

## Splitting Documents

```python
from langchain_core.documents import Document
```

```python
docs = [Document(
    page_content=text,
    metadata={'source': "Text on AI"}
)]
```

```python
docs
```

```python
splitter = CharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    separator=" "
)
```

```python
chunks = splitter.split_documents(docs)
chunks
```
